```python
import bs4
from langchain_community.document_loaders import WebBaseLoader

loader = WebBaseLoader(
    web_path=("https://lilianweng.github.io/posts/2023-06-23-agent/",),
    bs_kwargs=dict(
        parse_only = bs4.SoupStrainer(
            class_=("post-content", "post-title", "post-header")
        )
    )
)

doc = loader.load()
```

```python
from langchain.text_splitter import CharacterTextSplitter
from langchain_core.documents import Document

def chunk_fixed(docs, chunk_size = 512, overlap = 50) -> list[dict]:
    chunks = []
    step = chunk_size = overlap

    for doc in docs:
        words = doc.page_content.split()

        for i in range(0, len(words), step):
            chunk_words = " ".join(words[i: i+chunk_size])
            chunks.append(Document(
                page_content=chunk_words,
                metadata=doc.metadata
            ))

    return chunks

spliter = CharacterTextSplitter(separator = " ", chunk_size = 512, chunk_overlap = 50)
chunks_using_character_splitter = spliter.split_documents(doc)
chunks_using_custom_splitter = chunk_fixed(doc)
```
